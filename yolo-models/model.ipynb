```python
from PIL import Image
import os

for root, _, files in os.walk("animal_dataset/images"):
    for f in files:
        p = os.path.join(root, f)
        img = Image.open(p)
        if max(img.size) > 1920:
            img.thumbnail((1920, 1920))
            img.save(p)
            print(f"Resized {p}")
```

```python
from ultralytics import YOLO
```

```python

model = YOLO("yolov8n.pt")

model.train(
    data="data.yaml",
    epochs=25,
    imgsz=640,
    batch=16,
    device=0,
    name="animals_yolo",
)
```

```python
model.val()
```
