```python
from ultralytics import YOLO
```

```python
model = YOLO("best_models/best.pt")
```

```python
results = model.predict("test/bear.png", conf=0.05, save=True)

for r in results:
    for box in r.boxes:
        cls = int(box.cls[0])
        conf = float(box.conf[0])
        name = r.names[cls]
        print(name, conf)
```
